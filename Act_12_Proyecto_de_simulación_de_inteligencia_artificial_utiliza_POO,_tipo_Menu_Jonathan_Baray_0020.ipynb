```python
# 21/mayo/2026

# Farmacéutico:
# Registro sanitario, farmacia,
# nivel de inventario,
# especialidad en fórmulas,
# turnos y ventas del mes.

# ======================================================
# 💊 CBTis 128
# 📚 Elabora Proyectos de Programación Lógica
# 🤖 Especialidad: Inteligencia Artificial
# 👨‍🏫 Docente: Manuel Nava
# 👨‍🎓 Alumno: Jonathan Baray NL = 06
# 💻 Proyecto: Lista Farmacéutico IA
# ======================================================

# =========================
# 💊 Lista vacía
# =========================
lista_farmaceuticos = []

# =========================
# 📌 Menú principal
# =========================
while True:

    print("\n=========== MENÚ FARMACÉUTICO 💊 ===========")
    print("1. Agregar farmacéutico")
    print("2. Revisar inventario")
    print("3. Analizar ventas")
    print("4. Mostrar farmacéuticos")
    print("5. Salir")

    opcion = int(input("Selecciona una opción: "))

    # ==================================================
    # ➕ AGREGAR FARMACÉUTICO
    # ==================================================

    if opcion == 1:

        nombre = input("Ingrese nombre del farmacéutico: ")
        registro = input("Ingrese registro sanitario: ")
        farmacia = input("Ingrese nombre de farmacia: ")
        inventario = int(input("Ingrese nivel de inventario: "))
        especialidad = input("Ingrese especialidad en fórmulas: ")
        turno = input("Ingrese turno: ")
        ventas = float(input("Ingrese ventas del mes: "))

        lista_farmaceuticos.append([
            nombre,
            registro,
            farmacia,
            inventario,
            especialidad,
            turno,
            ventas
        ])

        print("\n✅ Farmacéutico agregado correctamente")

    # ==================================================
    # 📦 REVISAR INVENTARIO
    # ==================================================

    elif opcion == 2:

        inventario = int(input("Ingrese nivel de inventario: "))

        print("\n=========== ANÁLISIS IA ===========")

        if inventario >= 500:
            print("💊 Inventario alto y suficiente")

        elif inventario >= 200:
            print("📦 Inventario medio")

        else:
            print("⚠️ Inventario bajo, realizar pedido")

    # ==================================================
    # 💰 ANALIZAR VENTAS
    # ==================================================

    elif opcion == 3:

        ventas = float(input("Ingrese ventas del mes: "))

        print("\n=========== ANÁLISIS IA ===========")

        if ventas >= 50000:
            print("🏆 Ventas excelentes")

        elif ventas >= 20000:
            print("📈 Ventas aceptables")

        else:
            print("📉 Ventas bajas")

    # ==================================================
    # 📋 MOSTRAR FARMACÉUTICOS
    # ==================================================

    elif opcion == 4:

        print("\n==============================================================================================================================")
        print(f"{'NOMBRE':<20}{'REGISTRO':<18}{'FARMACIA':<20}{'INVENTARIO':<15}{'ESPECIALIDAD':<20}{'TURNO':<15}{'VENTAS':<15}")
        print("==============================================================================================================================")

        i = 0

        while i < len(lista_farmaceuticos):

            print(f"{lista_farmaceuticos[i][0]:<20}"
                  f"{lista_farmaceuticos[i][1]:<18}"
                  f"{lista_farmaceuticos[i][2]:<20}"
                  f"{lista_farmaceuticos[i][3]:<15}"
                  f"{lista_farmaceuticos[i][4]:<20}"
                  f"{lista_farmaceuticos[i][5]:<15}"
                  f"{lista_farmaceuticos[i][6]:<15}")

            i += 1

        print("==============================================================================================================================")

    # ==================================================
    # ❌ SALIR
    # ==================================================

    elif opcion == 5:

        print("\n💊 Programa finalizado")
        break

    # ==================================================
    # ⚠️ OPCIÓN INVÁLIDA
    # ==================================================

    else:

        print("❌ Opción no válida")
```
